```python
def intro():
    print("\n🌸 Welcome to 'The Crown of Aveloria' 🌸")
    name = input("Enter your name, brave one: ")
    print(f"\n{name}, you live a simple life... until one day a royal guard arrives.")
    return name


def royal_reveal(name):
    print("\n👑 The guard kneels before you.")
    print(f"'Princess {name}, you are the rightful heir to the throne!'")

    choice = input("\nDo you:\n1. Accept your destiny\n2. Refuse and run away\nEnter choice: ")

    if choice == "1":
        return "accept"
    else:
        return "run"


def palace_path(name):
    print("\n🏰 You arrive at the grand palace.")
    print("The kingdom is in danger. You must decide your approach.")

    choice = input("\nDo you:\n1. Train as a warrior ⚔️\n2. Rule with wisdom 📜\nEnter choice: ")

    if choice == "1":
        return warrior_path(name)
    else:
        return wisdom_path(name)


def warrior_path(name):
    print("\n⚔️ You train day and night, becoming a fierce warrior.")

    choice = input("\nAn enemy attacks! Do you:\n1. Fight bravely\n2. Protect your people first\nEnter choice: ")

    if choice == "1":
        print(f"\n🔥 {name}, you defeat the enemy and become a legendary Warrior Queen!")
    else:
        print(f"\n🛡️ {name}, you save your people and earn their eternal love as a Guardian Queen.")


def wisdom_path(name):
    print("\n📜 You study ancient texts and diplomacy.")

    choice = input("\nA rival kingdom challenges you. Do you:\n1. Negotiate peace\n2. Prepare for war\nEnter choice: ")

    if choice == "1":
        print(f"\n🕊️ {name}, you bring peace and become a Beloved Queen of Harmony.")
    else:
        print(f"\n⚖️ {name}, your tough decisions make you a Powerful but feared ruler.")


def runaway_path(name):
    print("\n🌙 You choose freedom over destiny and run away.")

    choice = input("\nDo you:\n1. Start a new life in another land\n2. Return later to reclaim your throne\nEnter choice: ")

    if choice == "1":
        print(f"\n🌼 {name}, you live peacefully, far from royal duties. A quiet but happy ending.")
    else:
        print(f"\n👑 {name}, you return stronger and reclaim your throne on your own terms!")


def main():
    name = intro()
    path = royal_reveal(name)

    if path == "accept":
        palace_path(name)
    else:
        runaway_path(name)

    print("\n✨ The End ✨")


if __name__ == "__main__":
    main()
```
